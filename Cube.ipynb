```python
import numpy as np
import matplotlib.pyplot as plt
```

# RGB Cube:
### هو فضاء يتم تمثيل كل لون كنقطقة داخل مكعب ثلاثي الأبعاد

# Cross-Section:
### هو تثبيت قيمة قناة واحدة وتغيير القناتيين الأُخريين للحصول على اقصى عدد من الألوان عند خلط القناتين
#### النتيجة عبارة عن مقطع ثنائي الأبعاد من مكعب RGB 

```python
R = 127

G, B = np.meshgrid(
    np.arange(256),
    np.arange(256)
)

image = np.zeros((256, 256, 3), dtype=np.uint8)

image[:, :, 0] = R
image[:, :, 1] = G
image[:, :, 2] = B

plt.imshow(image)
```
